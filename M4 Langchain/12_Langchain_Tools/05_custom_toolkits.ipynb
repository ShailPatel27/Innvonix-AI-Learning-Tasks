```python
from langchain_core.tools import tool

@tool
def add(a, b) -> int:
    """Add two numbers"""
    return a+b

@tool
def multiply(a, b) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
