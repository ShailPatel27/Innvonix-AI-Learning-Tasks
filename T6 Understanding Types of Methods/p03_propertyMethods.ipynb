## Property Methods

Description: Property methods allow you to manage the attributes of a class with getter, setter, and deleter methods using the @property decorator. This provides a way to add validation logic when getting or setting an attribute.

Challenge:

1. Define a class Person with a private attribute _age.
2. Create a property method age with a getter and setter to access and modify the _age attribute.
3. Instantiate the Person class, set the age, and print it.

```python
# Define class
class Person:

    def __init__(self):
        self._age = 0

    # Getter
    @property
    def age(self):
        return self._age

    # Setter
    @age.setter
    def age(self, value):
        self._age = value


# Create object
p = Person()

# Set age
p.age = 25

# Get age
print(p.age)
```
