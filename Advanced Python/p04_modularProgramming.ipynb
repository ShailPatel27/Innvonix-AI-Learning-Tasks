### Time to Execute

```python
import time
```

```python
def executionTime(func):
    
    def wrapper():
        start_time = time.time()
        
        func()
        
        end_time= time.time()
        
        print(f"time to execute: {end_time-start_time:.2f} seconds")
        
    return wrapper
```

```python
@executionTime
def greet():
    time.sleep(2)
    print("Hello World")
```

```python
greet()
```
