```python
from datetime import date
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
```

```python
current_year = date.today().year

c = Car("Swift", "Dzire", 2016)

print(f"Age of Car: {current_year-c.year}")
```
