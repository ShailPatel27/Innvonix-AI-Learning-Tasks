### Pickling

```python
import pickle


data = ["Shail", 21]

# Save object
with open("p08.pkl", "wb") as file:
    pickle.dump(data, file)


# Load object
with open("p08.pkl", "rb") as file:
    result = pickle.load(file)

print(result)
```

### Custom Exception

```python

class FileError(Exception):
    pass


def read_file(filename):

    try:
        with open(filename, "r") as file:
            print(file.read())

    except FileNotFoundError:
        raise FileError("File not found")


try:
    read_file("p08.txt")

except FileError as e:
    print(e)
```
