### Nested Class

```python
class Employee:

    class Address:
        def __init__(self, city):
            self.city = city

        def show(self):
            print("City:", self.city)


addr = Employee.Address("Ahmedabad")

addr.show()
```

### Class Methods

```python
class Calculator:

    @classmethod
    def add(cls, a, b):
        return a + b

    @classmethod
    def subtract(cls, a, b):
        return a - b

    @classmethod
    def multiply(cls, a, b):
        return a * b


print(Calculator.add(10, 5))
print(Calculator.subtract(10, 5))
print(Calculator.multiply(10, 5))
```
