### Abstract Method

```python
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def area(self):
        print("Area calculated")
```

### Simple Decorator

```python
def message(func):

    def wrapper():
        print("Before function")

        func()

        print("After function")

    return wrapper


@message
def hello():
    print("Hello")


hello()
```

### Time Function

```python
import time
```

```python
def exec(func):
    def wrapper():
        start = time.time()
        
        func()
        
        end = time.time()
        
        print(f"Execution time: {end-start:.2f}s")
        
    return wrapper
```

```python
@exec
def greet():
    for i in range(5000): i**i
    
greet()
```
