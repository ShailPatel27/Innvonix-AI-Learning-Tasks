## Introduction to Classes and Objects

Description

Task: Understand the concept of classes and objects in Python.

Task: Create a Python class with basic attributes.

Task: Instantiate objects from the created class.

Coding Challenge: Create a Python class representing a 'Car' with attributes like 'make', 'model', and 'year'. Instantiate two car objects and print their details.

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        
    def printDetails(self):
        print(f"Make: {self.make} | Model: {self.model} | Year: {self.year}")
```

```python
c1 = Car("Toyota", "Hyryder", 2026)
c2 = Car("Swift", "Dzire", 2016)

c1.printDetails()
c2.printDetails()
```
