## Static Methods

Description: Static methods do not take a self or cls parameter. They are defined using the @staticmethod decorator and are utility methods that perform a task in isolation.

Challenge:

1. Define a class MathUtils with a static method add_numbers that takes two numbers and returns their sum.
2. Call the static method without creating an instance of the class.

```python
# Define class
class MathUtils:

    # Static method
    @staticmethod
    def add_numbers(a, b):
        return a + b


# Call static method
result = MathUtils.add_numbers(5, 3)

print(result)
```
