### Arithmetic Operations

#### Integer

```python
a = 10
b = 20

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

#### Float

```python
a = 10.1
b = 20.2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

### String Manipulations

```python
fname = "Shail"
lname = "Patel"

print(f"Full Name: {fname + lname}")
print(fname[1:3])
print(fname.upper())
```
