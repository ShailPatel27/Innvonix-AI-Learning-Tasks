## Decorators in Classes

Description: Implement a class decorator to log method calls within a class.

Task: Create a decorator log_methods that logs the entry and exit of every method call within a class.

```python
# Class decorator
def log_methods(cls):

    # Loop through class methods
    for name, method in cls.__dict__.items():

        # Check if attribute is a method
        if callable(method):

            # Wrapper function
            def wrapper(*args, __method=method, __name=name, **kwargs):
                print(f"Entering {__name}")

                result = __method(*args, **kwargs)

                print(f"Exiting {__name}")

                return result

            # Replace original method
            setattr(cls, name, wrapper)

    return cls


@log_methods
class Test:

    def hello(self):
        print("Hello")

    def add(self, a, b):
        return a + b


t = Test()

t.hello()

print(t.add(2, 3))
```
