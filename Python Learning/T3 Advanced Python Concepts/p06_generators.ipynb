## Generators

Description: Generators are functions that return an iterable set of items, one at a time, in a special way. They use the yield keyword instead of return.

Challenge:

1. Write a generator function countdown that takes a number and counts down to zero.
2. Iterate over the generator and print each number.

```python
def countdown(number):
    while number >= 0:
        yield number
        number -= 1
```

```python
for i in countdown(5):
    print(i)    
```
