## Polymorphism

Description: Polymorphism allows methods to do different things based on the object it is acting upon, even though they share the same name.

Challenge:

1. Define a class Bike with a method start_engine that prints "Bike engine started".
2. Add a method start_engine to the Car class that prints "Car engine started".
3. Write a function start_vehicle that takes a vehicle object and calls its start_engine method.

```python
class Car:
    @staticmethod
    def start_engine():
        print("Car engine started")

class Bike:
    @staticmethod
    def start_engine():
        print("Bike engine started")

def start_vehicle(vehicle):
    vehicle.start_engine()


car1 = Car()
bike1 = Bike()


start_vehicle(car1)
start_vehicle(bike1)
```
