## Sales EDA

```python
import shutil

shutil.copy2("dirty_cafe_sales.csv", "01_dirty_cafe_sales.csv")
```

```python
import pandas as pd

df = pd.read_csv('01_dirty_cafe_sales.csv')

df
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors="coerce")
df["Total Spent"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"], errors="coerce")

for col in df:
    df[col] = df[col].replace(["UNKNOWN", "ERROR"], value=None)

df.dropna(inplace=True) #remove NaN
df
```

#### Dashboard

```python
print("Total Sales: ", df["Total Spent"].sum())
print("Total Items Sold: ", df["Quantity"].sum())
print("Total Quantity Sold: ", df["Quantity"].sum())
print("Average Transaction Value: ", round(float((df["Total Spent"].mean())), 2))
```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun",
          "Jul","Aug","Sep","Oct","Nov","Dec"]

df["Month"] = df["Transaction Date"].dt.strftime("%b")

df["Month"] = pd.Categorical(
    df["Month"],
    categories=months,
    ordered=True
)

monthly_sales = df.groupby("Month")["Total Spent"].sum()

monthly_sales.plot()

plt.show()
```

```python
df.groupby("Item")["Quantity"].sum().sort_values().plot(kind="barh")
```

```python
df["Payment Method"].value_counts().plot(kind="pie")
```

## Covid-19 Dataset

```python
import pandas as pd

df = pd.read_csv('02_covid_19_india.csv')
display(df)
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
import matplotlib.pyplot as plt

plt.title("Total cases over time")
plt.ylabel("Confirmed Cases")
daily_cases = df.groupby("Date")["Confirmed"].sum()
daily_cases.plot()
plt.show()
```

```python
plt.title("Deaths Over Time")

plt.ylabel("Deaths")
daily_deaths = df.groupby("Date")["Deaths"].sum()
daily_deaths.plot()
plt.show()
```

## Customer Segmentation

```python
import pandas as pd

df = pd.read_csv('03_customer_segmentation.csv')
display(df)
```

```python
df.dropna(inplace=True)
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

plt.title("Profession Distribution")
plt.ylabel("Count")
df["Profession"].value_counts().plot(kind="bar")

plt.show()
```

```python
plt.title("Age Distribution")
sns.histplot(df["Age"])

plt.show()
```

```python
sns.boxplot(
    x="Family_Size",
    y="Spending_Score",
    data=df
)

plt.show()
```

```python
plt.title("Spending Score Analysis")
df["Spending_Score"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.show()
```

```python
plt.title("Segmentation vs Spending")
sns.countplot(
    x="Segmentation",
    hue="Spending_Score",
    data=df
)

plt.show()
```
