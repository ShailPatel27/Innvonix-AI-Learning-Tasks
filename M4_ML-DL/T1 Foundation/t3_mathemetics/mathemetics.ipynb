### Probability Distributions

#### Normal Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=0,
    scale=1,
    size=1000
)

plt.hist(data)

plt.show()
```

#### Binomial Distribution

```python
data = np.random.binomial(
    n=10,
    p=0.5,
    size=1000
)

plt.hist(data)

plt.show()
```

#### Poisson Distribution

```python
data = np.random.poisson(
    lam=5,
    size=1000
)

plt.hist(data)

plt.show()
```
