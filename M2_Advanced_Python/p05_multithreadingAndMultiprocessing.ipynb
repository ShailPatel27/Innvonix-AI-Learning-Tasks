## Multithreading and Multiprocessing

Description

Coding Challenge: Implement a Python script using threads to download multiple images concurrently from a given list of URLs.

Coding Challenge: Create a program using multiprocessing to perform parallel computations on a large dataset.

```python
import urllib.request
import os
import threading


os.makedirs("images", exist_ok=True)

# Image URL
urls = [
    "https://picsum.photos/300",
    "https://picsum.photos/301",
    "https://picsum.photos/302"
    ]


def download(url, count): 
    urllib.request.urlretrieve(
        url,
        f"images/downloaded{count}.jpg"
    )
    
count = 1

for url in urls:
    t = threading.Thread(target=download, args=[url, count])
    t.start()
    print(f"Image {count} downloaded")
    count +=1
```
