## Database Handling

Description

Task: Introduction to database handling in Python.

Coding Challenge: Establish a connection to a MySQL database in Python and perform a basic query.

```python
import sqlite3


conn = sqlite3.connect("database.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER,
    name TEXT
)
""")

cursor.execute(
    "INSERT INTO users VALUES (?, ?)",
    (3, "Jeet")
)

conn.commit()

cursor.execute("SELECT * FROM users")

for row in cursor.fetchall():
    print(row)

conn.close()
```
