```python
from langchain_core.tools import tool
```

```python
# Step - 1 create a function

def multiply(a, b):
    """Multiply 2 numbers"""    # this is not optional
    return a*b
```

```python
# Step - 2 add type hints

def multiply(a: int, b: int):
    """Multiply 2 numbers"""
    return a*b
```

```python
# Step - 3 add tool decorator

@tool
def multiply(a: int, b: int):
    """Multiply 2 numbers"""
    return a*b
```

```python
result = multiply.invoke({"a":3, "b":5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```
