## Exception Handling using I/O

Description: Write a Python script that handles exceptions during file operations like reading from a non-existent file or writing to a read-only file.

Task: Try to read from a file non_existent.txt and handle the FileNotFoundError. Also, try to write to a file read_only.txt and handle the PermissionError.

### Handle FileNotFoundError

```python
try:
    with open("non_existent.txt", "r") as file:
        data = file.read()

except FileNotFoundError:
    print("File does not exist.")
```

### Handle PermissionError

```python
try:
    with open("read_only.txt", "w") as file:
        file.write("Hello")

except PermissionError:
    print("No permission to write to the file.")
```
