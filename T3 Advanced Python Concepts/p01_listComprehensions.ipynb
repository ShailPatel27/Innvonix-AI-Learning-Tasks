Description: List comprehensions provide a concise way to create lists. They consist of brackets containing an expression followed by a for clause, then zero or more for or if clauses.

Challenge:

1. Create a list of squares of numbers from 1 to 10 using a list comprehension.
2. Use a list comprehension to filter out even numbers from a list of numbers from 1 to 20.

```python
squares = [num ** 2 for num in range(1, 11)]

print(squares)
```

```python
even_numbers = [num for num in range(1, 20) if num % 2 == 0]
print(even_numbers)
```
