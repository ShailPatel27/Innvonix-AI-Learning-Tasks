## Variables and Assignment

Description: Write a Python script that demonstrates variable assignment and reassignment.

Task: Assign values to variables, perform operations with variables, and demonstrate variable reassignment.

### Variable assignment

```python
name = "Shail"
age = 20
salary = 5000000
bonus = 1
```

### Performing operations with variables

```python
total_salary = salary + bonus

print("Bonus:", bonus)
print("Total Salary:", total_salary)
```

### Variable reassignment

```python
age = 21
salary = total_salary

print("Updated Age:", age)
print("Updated Salary:", salary)
```
