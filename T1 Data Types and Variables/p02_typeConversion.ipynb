## Type Conversion

Create a Python script that demonstrates type conversion between different data types.

Task: Convert integers to floats, floats to integers, strings to integers, and vice versa.

```python
i: int = 10
f: float = 10.1
s: str = "20"
```

```python
a = float(i)
b = int(f)
c = int(s)
d = str(i)
```

```python
print(f"{a}: {type(a)}")
print(f"{b}: {type(b)}")
print(f"{c}: {type(c)}")
print(f"{d}: {type(d)}")
```
