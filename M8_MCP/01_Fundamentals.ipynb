## 1. What is MCP?

**MCP = Model Context Protocol.**

MCP is a standardized protocol that allows an AI application to connect to **external tools, data, and services**.

### Without MCP

```text
LLM
 ↓
Agent
 ↓
Python functions
 ↓
APIs / DB / Files
```

### With MCP

```text
LLM
 ↓
Agent
 ↓
MCP Client
 ↓
MCP Server
 ↓
Tools / Data / Services
```

### Key idea

> **MCP standardizes how AI applications communicate with external capabilities.**

## 2. Why do we need MCP?

Imagine several AI applications all need access to the same database.

### Without a standard

```text
App A → custom integration → Database
App B → custom integration → Database
App C → custom integration → Database
```

Each application can end up implementing its own integration.

### With MCP

```text
             MCP Server
                 │
              Database
                 ▲
       ┌─────────┼─────────┐
       │         │         │
     App A     App B     App C
   MCP Client MCP Client MCP Client
```

The applications communicate through the same standardized protocol.

### Main benefit

MCP reduces the need for every AI application to build a separate integration for every external capability.

## 3. MCP Architecture

There are three important terms:

### Host

The **AI application** that the user interacts with.

Examples can include:
- An AI coding application
- An AI assistant
- Your own AI application

The host manages the overall AI experience.

### Client

The MCP client is the component inside the host that communicates with MCP servers.

```text
Host
 │
 └── MCP Client
```

### Server

The MCP server exposes capabilities to the client.

For example:

```text
MCP Server
 ├── search_database()
 ├── read_file()
 ├── create_file()
 └── get_weather()
```

## 4. Complete Architecture

```text
┌─────────────────────┐
│        HOST         │
│                     │
│    LLM / Agent      │
│         │           │
│    MCP Client       │
└─────────┬───────────┘
          │
          │ MCP
          ▼
┌─────────────────────┐
│     MCP SERVER      │
│                     │
│  Tools / Resources  │
│       / Prompts     │
└─────────┬───────────┘
          │
          ▼
   External Systems
```

### Remember

- **Host** → AI application
- **Client** → communicates with MCP servers
- **Server** → exposes capabilities
- **MCP** → standardized communication protocol

## 5. MCP vs LangChain Tools

A LangChain tool might look like:

```python
from langchain_core.tools import tool

@tool
def calculate(a: int, b: int):
    return a + b
```

The tool exists directly inside your application:

```text
Agent
  │
  ▼
LangChain Tool
  │
  ▼
Python
```

With MCP:

```text
Agent
  │
  ▼
MCP Client
  │
  ▼
MCP Server
  │
  ▼
Tool
```

### Important

MCP does **not** replace LangChain.

They can work together:

```text
LangChain Agent
       │
       ▼
   MCP Client
       │
       ▼
   MCP Server
       │
       ▼
     Tools
```

Think of:

- **LangChain** → framework for building AI applications/agents
- **MCP** → protocol for connecting AI applications to external capabilities

## 6. MCP vs APIs

A REST API might expose endpoints such as:

```text
GET  /users
POST /users
GET  /orders
```

An MCP server can expose capabilities such as:

```text
get_user()
create_user()
get_orders()
```

MCP is not simply another REST API.

It provides a standardized, AI-oriented way to:

- Discover capabilities
- Describe tools
- Define tool schemas
- Call tools
- Exchange structured results
- Expose resources
- Expose prompts

### Simple distinction

```text
REST API
    ↓
General software ↔ software communication

MCP
    ↓
Standardized AI application ↔ capability communication
```

## 7. MCP Primitives — Preview

MCP servers can expose three important types of capabilities:

| Primitive | Purpose |
|---|---|
| **Tools** | Perform actions |
| **Resources** | Provide data/context |
| **Prompts** | Provide reusable prompt templates |

Examples:

```text
Tools
 ├── search_database()
 ├── create_file()
 └── calculate()

Resources
 ├── file://report.pdf
 └── database://users

Prompts
 ├── summarize_document
 └── analyze_code
```

We will study these properly in a later chapter.

## 8. Core Mental Model

The most important diagram from this chapter:

```text
                 AI APPLICATION
                       │
                 ┌─────▼─────┐
                 │ MCP Client│
                 └─────┬─────┘
                       │
                  MCP Protocol
                       │
                 ┌─────▼─────┐
                 │ MCP Server│
                 └─────┬─────┘
                       │
             ┌─────────┼─────────┐
             ▼         ▼         ▼
           Tools    Resources  Prompts
```

### One-line summary

> **Client communicates, Server exposes capabilities, and MCP standardizes the communication between them.**
