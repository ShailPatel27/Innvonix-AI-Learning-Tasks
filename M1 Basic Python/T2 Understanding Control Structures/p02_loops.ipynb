## Loops (for, while)

Description: Loops allow you to execute a block of code multiple times. The for loop iterates over a sequence (like a list, tuple, or string), while the while loop continues executing as long as a specified condition is true.

Challenge:

1. Write a function print_even_numbers that takes a limit as an argument.
2. Use a for loop to print all even numbers up to the limit.
3. Write another function countdown that takes a starting number as an argument.
4. Use a while loop to print a countdown from the starting number to 0.

```python
def print_even_numbers(limit):
    print("Even:")
    for i in range(0, limit):
        if(i % 2 == 0):
            print(i)
    print("")
```

```python
def countdown(starting_number):
    print("countdown")
    while starting_number >= 0:
        print(starting_number)
        starting_number -= 1
    print("")
```

```python
print_even_numbers(10)

countdown(10)
```
