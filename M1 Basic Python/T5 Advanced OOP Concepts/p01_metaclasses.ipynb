## Metaclasses

Description: Implement a metaclass that modifies the behavior of class creation.

Task: Create a metaclass SingletonMeta that ensures only one instance of a class can be created.

Metaclasses are classes that define classes. They are like a config for classes.

```python
# Defining a metaclass

class SingletonMeta(type):

    # Dictionary to store instances
    _instances = {}

    # Control object creation
    def __call__(cls, *args, **kwargs):

        # Create a new instance only if one does not exist
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)

        # Return the existing instance
        return cls._instances[cls]


# Class using the SingletonMeta metaclass

class Database(metaclass=SingletonMeta):

    def __init__(self):
        print("Database instance created")


# Creating objects

db1 = Database()
db2 = Database()

# Checking if both variables point to the same object

print(db1 is db2)
```
