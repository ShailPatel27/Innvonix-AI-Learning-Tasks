Description: A class is a blueprint for creating objects. An object is an instance of a class. Classes encapsulate data for the object.

Challenge:

1. Define a class Car with attributes make, model, and year.
2. Create an instance of the Car class.
3. Add a method display_info that prints out the details of the car.

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        
    def display_info(self):
        print("Car Details:")
        print("Make:", self.make)
        print("Model:", self.model)
        print("Year:", self.year)
        
c = Car("Toyota", "Hyryder", "2026")
c.display_info()
```
